## Instalación editable

La instalación editable permite que los cambios en `src/` se reflejen sin reinstalar el proyecto.

```python
import subprocess
import sys
from pathlib import Path

if not Path("pyproject.toml").is_file():
    raise FileNotFoundError(
        "Este cuaderno de desarrollo debe ejecutarse desde la raíz del repositorio."
    )
subprocess.check_call([sys.executable, "-m", "pip", "install", "-e", ".[dev,notebook]"])
print("Instalación editable de desarrollo completa.")
```
